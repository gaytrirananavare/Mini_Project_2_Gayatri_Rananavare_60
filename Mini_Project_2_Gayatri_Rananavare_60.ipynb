```python

from collections import deque

# Snakes: key = snake head, value = snake tail
snakes = {
    99: 54,
    70: 55,
    52: 42,
    25: 2,
    95: 72
}

# Ladders: key = ladder bottom, value = ladder top
ladders = {
    6: 25,
    11: 40,
    60: 85,
    46: 90,
    17: 69
}


def bfs_snake_ladder():
    start = 1
    destination = 100

    # Queue stores current position and path
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        position, path = queue.popleft()

        if position == destination:
            return path

        # Simulate dice values from 1 to 6
        for dice in range(1, 7):
            next_position = position + dice

            if next_position > destination:
                continue

            # Check for snake or ladder
            if next_position in snakes:
                next_position = snakes[next_position]
            elif next_position in ladders:
                next_position = ladders[next_position]

            if next_position not in visited:
                visited.add(next_position)
                queue.append(
                    (next_position, path + [next_position])
                )

    return None


# Main Program
print("=" * 45)
print("       SNAKE AND LADDER USING BFS")
print("=" * 45)

print("\nStarting Position:", 1)
print("Destination Position:", 100)

print("\nSnakes:", snakes)
print("Ladders:", ladders)

path = bfs_snake_ladder()

if path:
    print("\nShortest Path:")
    print(" -> ".join(map(str, path)))

    print("\nTotal Moves Required:", len(path) - 1)
else:
    print("\nDestination cannot be reached.")

```
